```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
from sklearn.datasets import make_regression
from sklearn.linear_model import LinearRegression
```

```python
X,y = make_regression(
    n_samples=100,
    n_features = 1,
    n_informative=1,
    n_targets=1,
    noise=20,
    random_state=42
)
```

```python
plt.scatter(X,y)

plt.xlabel("X")
plt.ylabel("y")
plt.title("Dataset")

plt.show()
```

```python
lr= LinearRegression()
```

```python
lr.fit(X,y)
```

```python
print("Coefficient(m):",lr.coef_)
print("intercept(b):",lr.intercept_)
```

```python
m = lr.coef_[0]
print("m=",m)
```

```python
b = -100

learning_rate = 0.001
epochs = 100
for i in range(epochs):
    #calculate gradient of b
    loss_slope = -2*np.sum(
        y-m*X.ravel()-b
    )
```

```python
#update b
b = b - (learning_rate*loss_slope)

print(
    "Epoch:",i+1,
    "Gradient:",loss_slope,
    "b:",b 
)
```

```python
print("\nFinal b:",b)
```

```python
y_pred = m*X+b
```

```python
#plot regresion
```

```python
plt.scatter(X,y)
plt.plot(X,y_pred)

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Linear Regression using Gradient Decent")
plt.show()
```

```python
#CREATE OUR OWN GRADIENT DESCENT REGRESSOR
```

```python
class GDRegressor:

    def __init__(self, learning_rate, epochs):

        self.m = 29.19
        self.b = -120

        self.lr = learning_rate
        self.epochs = epochs


    def fit(self, X, y):

        # Gradient Descent

        for i in range(self.epochs):

            # Calculate gradient of b

            loss_slope = -2 * np.sum(
                y - self.m * X.ravel() - self.b
            )

            # Update b

            self.b = self.b - (
                self.lr * loss_slope
            )

            print(
                "Epoch:", i + 1,
                "Gradient:", loss_slope,
                "b:", self.b
            )


    def predict(self, X):

        return self.m * X + self.b
```

```python
gd = GDRegressor(
    learning_rate =0.001,
    epochs = 100
)

```

```python
#train_model
```

```python
gd.fit(X,y)
```

```python
#check final parameter
```

```python
print("\nFinal m:",gd.m)
print("Final b :",gd.b)
```

```python
#make prediction
```

```python
y_pred= gd.predict(X)
```

```python
plt.scatter(X,y)
plt.plot(X,y_pred)
plt.xlabel("X")
plt.ylabel("y")

plt.show()
```
