```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
titanic=pd.read_csv('train.csv')
```

```python
titanic.head()
```

```python
flights= sns.load_dataset('flights')
```

```python
flights.head()
```

```python
iris= sns.load_dataset('iris')
```

```python
iris.head()
```

1.Scatterplot

```python
sns.scatterplot(data=iris,x='sepal_length',y='petal_length',hue= 
               'species')
plt.show()
```

```python
sns.scatterplot(data=iris,x='sepal_length',y='petal_length')
plt.show()
```

```python
sns.scatterplot(data=titanic,x='Age',y='Fare')
plt.show()
```

```python
sns.scatterplot(data=flights,x='year',y= 'passengers')
pltshow()
```

2.Barplot

Barplot = category+numerical

```python
sns.barplot(data=iris,x='species',y= 'petal_length')
plt.show()
```

```python
sns.barplot(data=flights,x= 'month',y='passengers')
plt.show()
```

3.Boxplot

```python
sns.boxplot(data=iris,x='species',y='petal_length')
plt.show()
```

Boxplot= category+numerical+distribution/outliers

```python
sns.countplot(data=titanic,x='Sex',hue='Survived')
plt.show()
```

```python
titanic.groupby('Sex')['Survived'].mean()
```

```python
sns.boxplot(data=titanic,x='Pclass',y='Fare')
plt.show()
```

```python
sns.boxplot(data=flights,x='month',y='passengers')
plt.show()
```
