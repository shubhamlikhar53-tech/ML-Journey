```python
#Gradient Descent From Scratch
```

```python
import numpy as np 


from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
```

```python
data = load_diabetes()

X = data.data
y= data.target
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=45)
```

```python
#Ridge regression Descent
```

```python
class MyRidgeGD:

    def __init__(
        self,
        epochs=500,
        learning_rate=0.005,
        alpha=0.001
    ):

        self.epochs = epochs
        self.learning_rate = learning_rate
        self.alpha = alpha

        self.coef_ = None
        self.intercept_ = None


    def fit(self, X_train, y_train):

        # Number of features
        n_features = X_train.shape[1]

        # Initialize coefficients
        self.coef_ = np.ones(
            n_features
        )

        # Initialize intercept
        self.intercept_ = 0

        # Create theta
        theta = np.insert(
            self.coef_,
            0,
            self.intercept_
        )

        # Add column of 1s
        X_train = np.insert(
            X_train,
            0,
            1,
            axis=1
        )

        # ------------------------------------
        # Gradient Descent
        # ------------------------------------

        for i in range(self.epochs):

            # Ridge gradient
            gradient = (
                X_train.T @ X_train @ theta
                - X_train.T @ y_train
            )

            # Regularization
            regularization = (
                self.alpha * theta
            )

            # Do NOT regularize intercept
            regularization[0] = 0

            # Total gradient
            gradient += regularization

            # Update theta
            theta = (
                theta
                - self.learning_rate * gradient
            )

        # Save parameters
        self.intercept_ = theta[0]

        self.coef_ = theta[1:]


    def predict(self, X_test):

        return (
            X_test @ self.coef_
            + self.intercept_
        )
```

```python
#create a model
```

```python
model = MyRidgeGD(
    epochs=500,
    learning_rate=0.005,
    alpha=0.001)
```

```python
#train
```

```python
model.fit(X_train,y_train)
```

```python
#Predict
```

```python
y_pred=model.predict(
    X_test
)
```

```python
#Evaluate
```

```python
score =r2_score(
    y_test,
    y_pred
)
```

```python
print("R2  Score:",score)

print("\nCoefficients:")
print(model.coef_)

print("\nIntercept:")
print(model.intercept_)
```
