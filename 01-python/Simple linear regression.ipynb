```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('placement.csv')
```

```python
df.head()
```

#Input and Output

```python
X = df[['cgpa']]
y = df['package']
```

Visualize data

```python
plt.scatter(df['cgpa'],df['package'])
plt.xlabel('CGPA')
plt.ylabel('package(in LPA)')
plt.show()
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test =train_test_split(X,y,test_size = 0.2,random_state=2)
```

Create a Model

```python
from sklearn.linear_model import LinearRegression
```

```python
lr = LinearRegression()
```

Train Model

```python
lr.fit(X_train,y_train)
```

Make Prediction

```python
prediction = lr.predict(X_test)

print(prediction)
```

#Predict for one student

```python
single_prediction = lr.predict(X_test.iloc[2].values.reshape(1,1))

print(single_prediction)
```

#Coefficient and intercept

```python
m = lr.coef_
b = lr.intercept_

print("Coefficient:",m)
print("Intercept:",b)
```

#Manual prediction

```python
cgpa = 8.58

predicted_package = m*cgpa+b

print("Predicted Package:",predicted_package)
```

#Regresion line

```python
X_sorted = np.sort(X,axis=0)

plt.scatter(X,y)
plt.plot(X_sorted,lr.predict(X_sorted))

plt.xlabel('CGPA')
plt.ylabel('Package (in LPA)')
plt.show()
```
